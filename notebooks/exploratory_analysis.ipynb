# Exploratory analysis

This notebook only explores the prepared MovieLens tables. Training, splitting, metrics, and LLM reranking live in `src/movie_recommender` and are run from `scripts/`.

Run `python scripts/prepare_data.py --data-dir ml-100k` before executing these cells.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from movie_recommender.config import AppConfig
from movie_recommender.data import load_processed
from movie_recommender.utils import project_root, resolve_path

config = AppConfig.from_yaml(project_root() / "configs" / "default.yaml")
data = load_processed(resolve_path(config.data.processed_dir))
data.summary
```

```python
import matplotlib.pyplot as plt

ratings = data.ratings
figure, axes = plt.subplots(1, 3, figsize=(14, 3.5))
ratings["rating"].value_counts().sort_index().plot(kind="bar", ax=axes[0], title="Ratings")
ratings.groupby("user_id").size().plot(kind="hist", bins=30, ax=axes[1], title="Ratings per user")
data.train.groupby("movie_id").size().plot(kind="hist", bins=30, ax=axes[2], title="Training ratings per movie")
figure.tight_layout()
plt.show()
```

The movie-popularity histogram uses the training split only. Validation and test interactions are held out by `make_chronological_split` and are not included in that histogram.
